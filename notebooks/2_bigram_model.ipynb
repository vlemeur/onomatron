# Bigram model

Count-based bigram over `data/processed/names.txt` (no training: the model is
the smoothed count matrix). The logic lives in the package — `onomatron.data`
(names, vocab) and `onomatron.bigram.BigramModel` — this notebook only wires it
up, plots the count matrix and samples names.


```python
import matplotlib.pyplot as plt

%matplotlib inline

from onomatron.bigram import BigramModel
from onomatron.data import Vocab, load_names
```

## Load names and build the vocab

The `'.'` sentinel is index 0 and marks both the beginning and the end of a
name — safe because the pipeline guarantees names.txt holds only latin letters
and spaces.


```python
names = load_names("../data/processed/names.txt")  # kernel cwd = notebooks/
vocab = Vocab(names)
len(names), len(vocab)
```

## Fit the bigram model


```python
model = BigramModel.fit(names, vocab)
model.counts.shape
```

## Count matrix


```python
N = model.counts
plt.figure(figsize=(24, 24))
plt.imshow(N, cmap="Blues")
for i in range(len(vocab)):
    for j in range(len(vocab)):
        chstr = vocab.itos[i] + vocab.itos[j]
        plt.text(j, i, chstr, ha="center", va="bottom", color="gray")
        plt.text(j, i, N[i, j].item(), ha="center", va="top", color="gray")
plt.axis("off")
```

## Sample names


```python
for name in model.sample(n=100, seed=135798):
    print(name)
```

## Evaluation: mean NLL per bigram


```python
print(f"mean NLL per bigram: {model.nll(names):.4f}")
```
